# Project FORESIGHT — Exploratory Data Analysis
This notebook profiles sales demand trends, seasonality, ABC categorization, and promotional lifts for NorthBay Living.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/analysis_ready.csv', parse_dates=['week_start'])
print(df.head())
```

### ABC Classification Analysis

```python
sku_sales = df.groupby('sku_id')['revenue'].sum().sort_values(ascending=False).reset_index()
sku_sales['cum_pct'] = sku_sales['revenue'].cumsum() / sku_sales['revenue'].sum()
sku_sales['class'] = np.where(sku_sales['cum_pct'] <= 0.80, 'A', np.where(sku_sales['cum_pct'] <= 0.95, 'B', 'C'))
print(sku_sales['class'].value_counts())
```
